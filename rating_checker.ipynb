# Rating checker

```python
# name, star rating, number of reviews
OPTIONS = [
    ("Helmet A", 5.0, 10),
    ("Helmet B", 4.6, 1800),
]

m = 4.1      # typical score
C = 10       # pretend reviews
NOISE = 1.3  # spread of one review
```

Fair score = `(C·m + rating·n) / (C + n)`

```python
import numpy as np

rng = np.random.default_rng(1)
draws = []
for name, rating, n in OPTIONS:
    score = (C * m + rating * n) / (C + n)
    draws.append(np.clip(rng.normal(score, NOISE / np.sqrt(C + n), 20_000), 1, 5))
    print(f"{name}: {rating:.1f} from {n:,} reviews → fair score {score:.2f}")

best = np.argmax(draws, axis=0)
print()
for i, (name, _, _) in enumerate(OPTIONS):
    print(f"{name} is really the best in {np.mean(best == i):.0%} of plausible worlds")
```
